```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def load_dataset():
 #Load the dataset.
 dataset = pd.read_csv("../BostonHousing.csv")
 dataset.info()
 return dataset


```

```python
def summary_stat():
#Explore the data structure, types, and summary statistics.
    ds =load_dataset()
    print("\nData types:")
    print(ds.dtypes)

    print(ds.dtypes.value_counts())

    print(ds.describe())

    print("\nColumns summary:")
    print(ds.describe())

    print("\n Missing Values:")
    print(ds.isnull().sum())

    print("\nNumber of duplicated rows:")
    print(ds.duplicated().sum())

#summary_stat()
```

```python
def bivariate_relationship():
# Include descriptive statistics and appropriate visualizations
# histplot for df.columns
 ds = load_dataset()
 plt.figure(figsize= (15,20), constrained_layout = True)
 numerical_features = [col for col in ds.columns
                       if col not in ['chas', 'medv']]
 for i,var in enumerate(numerical_features, 1):
   descriptive_stats = ds[var].describe()
   ax = plt.subplot(5,3,i)
   sns.scatterplot(x = ds[var], y = ds['medv'], alpha = 0.6, ax = ax)
   
   plt.title(f'{var} vs MEDV')
   plt.xlabel(var, labelpad= 10)
   plt.ylabel('Median Value ($1000s)', labelpad= 10)
plt.tight_layout(pad = 3, h_pad = 2, w_pad = 2)
plt.show()


```

```python
def outliers_and_missingvalues_detection():
    df = load_dataset()
    plt.figure(figsize=(10, 6))
    sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
    plt.title("Missing Values Heatmap")
    plt.show()

    plt.figure(figsize=(15, 8))
    sns.violinplot(data=df.select_dtypes(include=[np.number]))
    plt.xticks(rotation=45)
    plt.title("Violin Plots Showing Distributions and Outliers")
    plt.show()
#outliers_and_missingvalues_detection()
```
